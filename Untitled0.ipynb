```python


from tensorflow.keras.datasets import fashion_mnist
import matplotlib.pyplot as plt
import numpy as np
from tensorflow import keras

labels = ["T-shirt/top", "Trouser", "Pullover", "Dress",
          "Coat", "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]


(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

plt.figure(figsize=(14, 8))
for i in range(20):
    plt.subplot(4, 5, i + 1)
    plt.imshow(X_train[i], cmap="binary")
    plt.title(labels[y_train[i]])
    plt.axis("off")
plt.tight_layout()
plt.show()


X_train, X_test = X_train / 255.0, X_test / 255.0



def entrenar_modelo(neuronas=128, capas=1, epochs=10):
    capas_modelo = [keras.layers.Flatten(input_shape=(28, 28))]
    for _ in range(capas):
        capas_modelo.append(keras.layers.Dense(neuronas, activation='relu'))
    capas_modelo.append(keras.layers.Dense(10, activation='softmax'))

    model = keras.Sequential(capas_modelo)
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    history = model.fit(
        X_train, y_train,
        epochs=epochs,
        validation_data=(X_test, y_test),
        verbose=0
    )
    _, acc = model.evaluate(X_test, y_test, verbose=0)
    return model, history, acc

print("=" * 50)
print("MEJORA 1: Comparación de épocas (10 vs 50)")
print("=" * 50)

_, hist_10, acc_10 = entrenar_modelo(epochs=10)
_, hist_50, acc_50 = entrenar_modelo(epochs=50)

print(f"Precisión con 10 épocas: {acc_10:.4f}")
print(f"Precisión con 50 épocas: {acc_50:.4f}")

plt.figure(figsize=(10, 4))
plt.plot(hist_10.history['val_accuracy'], label='10 épocas')
plt.plot(hist_50.history['val_accuracy'], label='50 épocas')
plt.xlabel("Época")
plt.ylabel("Precisión de validación")
plt.title("Comparación: 10 vs 50 épocas")
plt.legend()
plt.tight_layout()
plt.show()


print("\n" + "=" * 50)
print("MEJORA 2: Comparación de neuronas en capa oculta")
print("=" * 50)

resultados_neuronas = {}
for n in [64, 128, 256]:
    _, _, acc = entrenar_modelo(neuronas=n, epochs=10)
    resultados_neuronas[n] = acc
    print(f"Neuronas: {n} → Precisión: {acc:.4f}")

mejor_config = max(resultados_neuronas, key=resultados_neuronas.get)
print(f"\nMejor configuración: {mejor_config} neuronas "
      f"(precisión: {resultados_neuronas[mejor_config]:.4f})")

plt.figure(figsize=(7, 4))
plt.bar([str(k) for k in resultados_neuronas.keys()],
        resultados_neuronas.values(), color=['#378ADD', '#1D9E75', '#D85A30'])
plt.xlabel("Número de neuronas")
plt.ylabel("Precisión")
plt.title("Precisión según cantidad de neuronas en capa oculta")
plt.ylim(0.8, 1.0)
plt.tight_layout()
plt.show()



print("\n" + "=" * 50)
print("MEJORA 3: Una capa oculta vs dos capas ocultas")
print("=" * 50)

_, _, acc_1capa = entrenar_modelo(neuronas=128, capas=1, epochs=10)
_, _, acc_2capas = entrenar_modelo(neuronas=128, capas=2, epochs=10)

print(f"Precisión con 1 capa oculta (128):      {acc_1capa:.4f}")
print(f"Precisión con 2 capas ocultas (128+128): {acc_2capas:.4f}")

model_2capas = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])
model_2capas.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
model_2capas.fit(X_train, y_train, epochs=10,
                 validation_data=(X_test, y_test), verbose=0)



print("\n" + "=" * 50)
print("MEJORA 4: Predicciones vs valores reales (5 imágenes)")
print("=" * 50)

indices = np.random.choice(len(X_test), 5, replace=False)
predicciones = model_2capas.predict(X_test[indices])

plt.figure(figsize=(14, 4))
for i, idx in enumerate(indices):
    pred_label = labels[np.argmax(predicciones[i])]
    real_label = labels[y_test[idx]]
    color = 'green' if pred_label == real_label else 'red'

    plt.subplot(1, 5, i + 1)
    plt.imshow(X_test[idx], cmap="binary")
    plt.title(f"Real: {real_label}\nPred: {pred_label}", color=color, fontsize=9)
    plt.axis("off")

plt.suptitle("Predicciones vs Valores Reales (verde=correcto, rojo=incorrecto)",
             fontsize=11)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# OPERACIÓN 02: Regresión Lineal Simple - Temperatura vs Ventas
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

# ----------------------------
# Datos originales (ficticios)
# ----------------------------
temperatura = np.array([15, 16, 18, 20, 21, 23, 25, 27, 30, 32])
ventas      = np.array([500, 520, 560, 580, 600, 640, 680, 700, 760, 800])

X_train, X_test, y_train, y_test = train_test_split(
    temperatura.reshape(-1, 1), ventas,
    test_size=0.2, random_state=42
)

model_original = LinearRegression()
model_original.fit(X_train, y_train)
r2_original = model_original.score(X_test, y_test)

plt.figure(figsize=(7, 4))
plt.scatter(temperatura, ventas, color="blue", label="Datos reales")
plt.plot(temperatura,
         model_original.predict(temperatura.reshape(-1, 1)),
         color="red", label="Línea de regresión")
plt.xlabel("Temperatura (°C)")
plt.ylabel("Ventas")
plt.title("Regresión Lineal Simple: Temperatura vs Ventas (datos originales)")
plt.legend()
plt.tight_layout()
plt.show()

print(f"R² con datos originales: {r2_original:.4f}")


# ============================================================
# MEJORA 5: Nuevos datos de temperatura y ventas
# ============================================================

print("\n" + "=" * 50)
print("MEJORA 5: Análisis con datos ampliados")
print("=" * 50)

# Datos ampliados: se agregan nuevas observaciones
temperatura_nueva = np.array([
    15, 16, 18, 20, 21, 23, 25, 27, 30, 32,   # originales
    10, 12, 35, 38, 40, 22, 28, 33, 17, 26    # nuevos
])
ventas_nueva = np.array([
    500, 520, 560, 580, 600, 640, 680, 700, 760, 800,  # originales
    430, 460, 820, 870, 910, 630, 720, 790, 540, 670   # nuevos
])

X_train_n, X_test_n, y_train_n, y_test_n = train_test_split(
    temperatura_nueva.reshape(-1, 1), ventas_nueva,
    test_size=0.2, random_state=42
)

model_nuevo = LinearRegression()
model_nuevo.fit(X_train_n, y_train_n)
r2_nuevo = model_nuevo.score(X_test_n, y_test_n)

# Gráfica comparativa
temp_range = np.linspace(temperatura_nueva.min(), temperatura_nueva.max(), 100)

plt.figure(figsize=(10, 5))
plt.scatter(temperatura, ventas,
            color="blue", label="Datos originales", zorder=5)
plt.scatter(temperatura_nueva[10:], ventas_nueva[10:],
            color="orange", marker="^", label="Datos nuevos", zorder=5)
plt.plot(temp_range,
         model_original.predict(temp_range.reshape(-1, 1)),
         color="red", linestyle="--", label=f"Regresión original (R²={r2_original:.4f})")
plt.plot(temp_range,
         model_nuevo.predict(temp_range.reshape(-1, 1)),
         color="green", label=f"Regresión ampliada (R²={r2_nuevo:.4f})")
plt.xlabel("Temperatura (°C)")
plt.ylabel("Ventas")
plt.title("Comparación: Regresión con datos originales vs datos ampliados")
plt.legend()
plt.tight_layout()
plt.show()

print(f"R² con datos originales: {r2_original:.4f}")
print(f"R² con datos ampliados:  {r2_nuevo:.4f}")
print(f"\nPendiente original:  {model_original.coef_[0]:.4f}")
print(f"Pendiente ampliada:  {model_nuevo.coef_[0]:.4f}")
print(f"\nIntercepto original: {model_original.intercept_:.4f}")
print(f"Intercepto ampliado: {model_nuevo.intercept_:.4f}")
```

```python
# ============================================================
# OPERACIÓN 01: Red Neuronal con TensorFlow - Fashion MNIST
# ============================================================

from tensorflow.keras.datasets import fashion_mnist
import matplotlib.pyplot as plt
import numpy as np
from tensorflow import keras

labels = ["T-shirt/top", "Trouser", "Pullover", "Dress",
          "Coat", "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]

# ----------------------------
# Paso 2: Cargar y explorar el dataset
# ----------------------------
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

plt.figure(figsize=(14, 8))
for i in range(20):
    plt.subplot(4, 5, i + 1)
    plt.imshow(X_train[i], cmap="binary")
    plt.title(labels[y_train[i]])
    plt.axis("off")
plt.tight_layout()
plt.show()

# ----------------------------
# Paso 3: Normalización de datos
# ----------------------------
X_train, X_test = X_train / 255.0, X_test / 255.0


# ============================================================
# MEJORA 1: Comparar 10 épocas vs 50 épocas
# ============================================================

def entrenar_modelo(neuronas=128, capas=1, epochs=10):
    capas_modelo = [keras.layers.Flatten(input_shape=(28, 28))]
    for _ in range(capas):
        capas_modelo.append(keras.layers.Dense(neuronas, activation='relu'))
    capas_modelo.append(keras.layers.Dense(10, activation='softmax'))

    model = keras.Sequential(capas_modelo)
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    history = model.fit(
        X_train, y_train,
        epochs=epochs,
        validation_data=(X_test, y_test),
        verbose=0
    )
    _, acc = model.evaluate(X_test, y_test, verbose=0)
    return model, history, acc

print("=" * 50)
print("MEJORA 1: Comparación de épocas (10 vs 50)")
print("=" * 50)

_, hist_10, acc_10 = entrenar_modelo(epochs=10)
_, hist_50, acc_50 = entrenar_modelo(epochs=50)

print(f"Precisión con 10 épocas: {acc_10:.4f}")
print(f"Precisión con 50 épocas: {acc_50:.4f}")

plt.figure(figsize=(10, 4))
plt.plot(hist_10.history['val_accuracy'], label='10 épocas')
plt.plot(hist_50.history['val_accuracy'], label='50 épocas')
plt.xlabel("Época")
plt.ylabel("Precisión de validación")
plt.title("Comparación: 10 vs 50 épocas")
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# MEJORA 2: Comparar configuraciones de neuronas (64, 128, 256)
# ============================================================

print("\n" + "=" * 50)
print("MEJORA 2: Comparación de neuronas en capa oculta")
print("=" * 50)

resultados_neuronas = {}
for n in [64, 128, 256]:
    _, _, acc = entrenar_modelo(neuronas=n, epochs=10)
    resultados_neuronas[n] = acc
    print(f"Neuronas: {n} → Precisión: {acc:.4f}")

mejor_config = max(resultados_neuronas, key=resultados_neuronas.get)
print(f"\nMejor configuración: {mejor_config} neuronas "
      f"(precisión: {resultados_neuronas[mejor_config]:.4f})")

plt.figure(figsize=(7, 4))
plt.bar([str(k) for k in resultados_neuronas.keys()],
        resultados_neuronas.values(), color=['#378ADD', '#1D9E75', '#D85A30'])
plt.xlabel("Número de neuronas")
plt.ylabel("Precisión")
plt.title("Precisión según cantidad de neuronas en capa oculta")
plt.ylim(0.8, 1.0)
plt.tight_layout()
plt.show()


# ============================================================
# MEJORA 3: Agregar segunda capa oculta
# ============================================================

print("\n" + "=" * 50)
print("MEJORA 3: Una capa oculta vs dos capas ocultas")
print("=" * 50)

_, _, acc_1capa = entrenar_modelo(neuronas=128, capas=1, epochs=10)
_, _, acc_2capas = entrenar_modelo(neuronas=128, capas=2, epochs=10)

print(f"Precisión con 1 capa oculta (128):      {acc_1capa:.4f}")
print(f"Precisión con 2 capas ocultas (128+128): {acc_2capas:.4f}")

# Modelo con 2 capas para usar en predicciones
model_2capas = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])
model_2capas.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
model_2capas.fit(X_train, y_train, epochs=10,
                 validation_data=(X_test, y_test), verbose=0)


# ============================================================
# MEJORA 4: Mostrar predicciones de 5 imágenes del conjunto de prueba
# ============================================================

print("\n" + "=" * 50)
print("MEJORA 4: Predicciones vs valores reales (5 imágenes)")
print("=" * 50)

indices = np.random.choice(len(X_test), 5, replace=False)
predicciones = model_2capas.predict(X_test[indices])

plt.figure(figsize=(14, 4))
for i, idx in enumerate(indices):
    pred_label = labels[np.argmax(predicciones[i])]
    real_label = labels[y_test[idx]]
    color = 'green' if pred_label == real_label else 'red'

    plt.subplot(1, 5, i + 1)
    plt.imshow(X_test[idx], cmap="binary")
    plt.title(f"Real: {real_label}\nPred: {pred_label}", color=color, fontsize=9)
    plt.axis("off")

plt.suptitle("Predicciones vs Valores Reales (verde=correcto, rojo=incorrecto)",
             fontsize=11)
plt.tight_layout()
plt.show()
```
